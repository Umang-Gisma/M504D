##**If Else**

###**Leap year calculation**

```python
#Leap year calculation
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("This is a leap year")
else:
    print("This is not a leap year")
```

###**Students grade calculation**

```python
marks = int(input("Enter your marks: "))

if marks >= 90:
    print("Grade: A")
elif marks >= 80 and marks < 90:
    print("Grade: B")
elif marks >= 70 and marks < 80:
    print("Grade: C")
elif marks >= 60 and marks < 70:
    print("Grade: D")
else:
    print("Fail")
```

```python
marks = int(input("Enter your marks: "))

if marks >= 60:
    if marks >= 70:
        if marks >= 80:
            if marks >= 90:
                print("Grade: A")
            else:
                print("Grade: B")
        else:
            print("Grade: C")
    else:
        print("Grade: D")
else:
    print("Fail")
```

###**Calculation of average marks**

```python
sub1 = int(input("Enter marks for Subject 1: "))
sub2 = int(input("Enter marks for Subject 2: "))
sub3 = int(input("Enter marks for Subject 3: "))

if 0 <= sub1 <= 100 and 0 <= sub2 <= 100 and 0 <= sub3 <= 100:
    average = (sub1 + sub2 + sub3) / 3
    print("Average marks:", average)
else:
    print("Please enter marks between 0 and 100")
```

###Break

```python
# Print numbers 1 to 10, but stop if number is 5
for i in range(1, 11):
    if i == 5:
        print("Stopping the loop at", i)
        break  # exits the loop
    print(i)
```

```python
tNumber = int(input("Enter your desired table number: "))
sRange = int(input("Range starting from: "))
eRange = int(input("Range ending at: "))
brk = int(input("Enter the number to break the loop: "))

for i in range(sRange, eRange + 1):
    if i == brk:
        print("Stopping the loop at", i)
        break  # stop
    print(tNumber, "x", i, "=", tNumber * i)
```

###Continue

```python
# Print numbers 1 to 10, but skip number 5
for i in range(1, 11):
    if i == 5:
        print("Skipping", i)
        continue  # skip this iteration
    print(i)
```

```python
tNumber = int(input("Enter your desired table number: "))
sRange = int(input("Range starting from: "))
eRange = int(input("Range ending at: "))
skp = int(input("Enter the number to skip the loop: "))

for i in range(sRange, eRange + 1):
    if i == skp:
        print("Skipping the loop at", i)
        continue  # skip
    print(tNumber, "x", i, "=", tNumber * i)
```

##**While Loop**

###Intro

```python
i = 1  # starting point

while i <= 5:  # condition
    print(i)
    i += 1  # increment
```

###Collections and its functions

```python
# Start with a list of numbers
numbers = [1, 2, 3, 4, 5]
index = 0  # for the while loop

print("Original list:", numbers)
print("Length of list:", len(numbers))

# Using while loop to print each number
print("\nPrinting numbers using while loop:")
while index < len(numbers):
    print(numbers[index])
    index += 1

# Adding a new number
numbers.append(6)
print("\nAfter appending 6:", numbers)

# Inserting a number at index 2
numbers.insert(2, 10)
print("After inserting 10 at index 2:", numbers)

# Removing the last number
numbers.pop()
print("After popping the last element:", numbers)
```
