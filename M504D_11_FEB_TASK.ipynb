#**Count Vowel Characters**

Write a Python function that takes a string and returns the number of vowel characters in that string. Vowel characters in English are "a", "e", "i", "o", and "u". For example, given the string "abeecdA", the function should return 4.

```python
def count_vowels(s):
    vowels = "aeiou"       # define a string containing all vowel characters
    count = 0              # initialize a counter to keep track of vowels

    for char in s.lower():
        if char in vowels:   # check if the character is a vowel
            count += 1       # if it is, increase the count by 1

    return count

print(count_vowels("abeecdA"))
```

#M**ultiply the List Elements**

Write a Python function that takes a list of values as input and returns the product of all the numbers in the list. For example, given the input list [1, 3, 8], the function should return 24.

```python
def multiply_list(lst):
    product = 1            # Initialize a variable to store the product

    for num in lst:
        product *= num     # Multiply the current product by the number

    return product

print(multiply_list([1, 3, 8]))
```

#**Check if a String is Palindrome**

Write a Python function that takes a string and returns True if it is palindrome. A palindrome is a word, number, phrase, or other sequence of characters which reads the same backward as forward. For example, given the string "abc" the function should return False. But, given the string "aba", the function should return True.

```python
def is_palindrome(value):
    left = 0                    # Beginning of the string
    right = len(value) - 1      # End of the string

    while left < right:
        if value[left] != value[right]:  # Compare characters at both ends
            return False                 # If they are different, it is not

        left += 1               # Move the left pointer one step to the right
        right -= 1              # Move the right pointer one step to the left

    return True

print(is_palindrome("abc"))
```

#**Count the Unique Digits in an Integer**

Write a Python function that takes a positive integer and returns the number of unique digits in that integer. For example, given the integer 12315, the function should return 4.

```python
def count_unique_digits(n):
    digits_seen = []  # list to store unique digits

    for digit in str(n):  # iterate over each digit
        if digit not in digits_seen:  # check if we’ve seen it before
            digits_seen.append(digit)  # add it if it’s new

    return len(digits_seen)  # return the count of unique digits
print(count_unique_digits(1122333))
```

#**Check a Number is Prime**

Write a Python function that takes a number and checks the number is prime or not. It should return True for being prime and False otherwise. Note that a prime number (or a prime) is a natural number greater than 1 and that has no positive divisors other than 1 and itself. For example, the function should return True for the given number 7 and False for the given number 10.

```python
def is_prime(n):
    # Any number less than 2 is not prime
    if n < 2:
        return False

    # Check for divisors from 2 up to n-1
    for i in range(2, n):
        if n % i == 0:  # If n is divisible by i, it is not prime
            return False

    # If no divisors were found, n is prime
    return True


print(is_prime(7))
print(is_prime(10))
```

#**Sort Hyphened-Seperated Words**

Write a Python function that takes a sequence of words that are separated by hyphens as input and returns the words in a hyphen-separated sequence after sorting the words alphabetically. For example, given the string "green-red-yellow-black-white", the function should return "black-green-red-white-yellow".

```python
def sort_hyphen_words(s):

    # Split the string into a list of words using hyphen as separator
    words = s.split("-")

    # Sort the list of words alphabetically
    words.sort()

    # Join the sorted words back into a string with hyphens
    sorted_string = "-".join(words)

    return sorted_string

print(sort_hyphen_words("green-red-yellow-black-white"))
```

#**Count the Characters**

Write a Python function that takes a string as input and returns a dictionary that stores the frequency of each character in the input string. For example, given the string "abca", the function should return {'a': 2, 'b': 1, 'c': 1}.

```python
def char_frequency(s):
    freq = {}  # create an empty dictionary to store character counts

    # Loop through each character in the string
    for char in s:
        if char in freq:
            freq[char] += 1  # If the character is already in the dictionary, increase its count
        else:
            freq[char] = 1   # If the character is not in the dictionary, add it with count 1

    return freq  # return the dictionary with frequencies

print(char_frequency("abcaa"))
```

#**Fibonacci Sequence**

Compute the Fibonacci Sequence The Fibonacci Sequence is computed based on the following formula:

f(n)=0 if n=0;
f(n)=1 if n=1;
f(n)=f(n-1)+f(n-2) if n>1

Write a Python function that computes the value of f(n) with a given n as input. For example, given 7, the function should return 13.

```python
def fibonacci(n):
    # Handle the base cases
    if n == 0:
        return 0
    elif n == 1:
        return 1

    # For n > 1, compute Fibonacci using a loop
    a = 0  # f(0)
    b = 1  # f(1)

    for _ in range(2, n + 1):
        c = a + b  # f(n) = f(n-1) + f(n-2)
        a = b      # Move forward: a becomes previous b
        b = c      # Move forward: b becomes current Fibonacci

    return b  # b now holds f(n)

print(fibonacci(7))
```
