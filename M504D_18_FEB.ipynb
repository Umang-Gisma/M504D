#**Class and Object**

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def information(self):
    print("Hello my name is " + self.name + " and my age is " + str(self.age))

p1 = Person("Umang",22)
print(p1.name)
print(p1.age)
p1.information()
```

```python
from datetime import date

class Person:
    def __init__(self, name, birth_year):
        self.name = name
        self.birth_year = int(birth_year)

    def calculate_age(self):
        current_year = date.today().year
        return current_year - self.birth_year

p1 = Person("Umang", "2004")

print("Name:", p1.name)
print("Current Age:", p1.calculate_age())
```

#**Built-in attribute**

```python
class Person:
    def __init__(self, name):
        self.name = name

p1 = Person("Umang")

print("Class Person was defined in module:", Person.__module__)
print("Class Person was defined in dict:", Person.__dict__)
```

#**Class-side data and method**

```python
class Person:
    count = 0

    def __init__(self, name, age):
        self.name = name
        self.age = age
        Person.count += 1

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")

    @classmethod
    def total_people(cls):
        print(f"Total People created: {cls.count}")

p1 = Person("Umang", 22)
p2 = Person("Sanket", 25)
p3 = Person("Shubham", 30)

p1.display_info()
p2.display_info()
p3.display_info()

Person.total_people()
```

#**Instance and Static Methods**

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print(f"Hello, my name is {self.name}")

    @staticmethod
    def welcome_message():
        print("Welcome! This is a static message.")

# Object method
p1 = Person("Umang")
p1.greet()

# Static method
Person.welcome_message()
```

#**Encapsulation with Private Attributes**

```python
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    # Getter for name
    def get_name(self):
        return self.__name

    # Setter for name
    def set_name(self, name):
        self.__name = name

    # Getter for age
    def get_age(self):
        return self.__age

    # Setter for age (with validation)
    def set_age(self, age):
        if age >= 0:
            self.__age = age
        else:
            print("Age cannot be negative!")

    # Method to display info
    def display_info(self):
        print(f"Name: {self.__name}, Age: {self.__age}")


# Create object
p1 = Person("Umang", 22)

# Access private attributes using getter
print(p1.get_name())
print(p1.get_age())

# Modify private attributes using setter
p1.set_name("Rohit")
p1.set_age(25)

# Display updated info
p1.display_info()
```
