#**Student Class**

**Define a student class with the followings:**

A first_name attribute. A last_name attribute. An number attribute. A courses attribute that stores a dictionary of passed course objects and their mark. An init method. A str method. A get_gpa method. A set_course_mark method.

```python
class Student:
  def __init__(self, first_name, last_name, number, course):
    self.first_name = first_name
    self.last_name = last_name
    self.number = number
    self.course = course
  def __str__(self):
    return (f"Student name :{self.first_name} {self.last_name}\nStudent numer: {self.number}\nStudent courses: {self.course}")
  def get_gpa(self):
    total_marks = 0
    for i in self.course.values():
      total_marks += i
    return total_marks/len(self.course)

s1 = Student("Umang","Pipaliya","123",{"M504":78,"M505":85})

print(s1)
print(s1.get_gpa())
```
